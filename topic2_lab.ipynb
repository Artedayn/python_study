```python
print("Введите список товаров, которые вы хотите купить (через запятую):")
items = input().strip().split(',')

items = [item.strip() for item in items]

stores = {
    "Магазин 1": {"хлеб": 50, "молоко": 80, "сыр": 200, "яблоки": 100},
    "Магазин 2": {"хлеб": 45, "молоко": 85, "сыр": 190, "яблоки": 120},
    "Магазин 3": {"хлеб": 55, "молоко": 75, "сыр": 210, "яблоки": 90},
}

total_costs = {}

for store, prices in stores.items():
    total_cost = 0
    for item in items:
        if item in prices:
            total_cost += prices[item]
        else:
            print(f"Товар '{item}' отсутствует в {store}.")
            total_cost = None
            break
    if total_cost is not None:
        total_costs[store] = total_cost

print("\nОбщая стоимость покупок в каждом магазине:")
for store, cost in total_costs.items():
    print(f"{store}: {cost} рублей")

if total_costs:
    cheapest_store = min(total_costs, key=total_costs.get)
    cheapest_cost = total_costs[cheapest_store]
    print(f"\nСамый выгодный магазин: {cheapest_store} (стоимость: {cheapest_cost} рублей)")

    most_expensive_cost = max(total_costs.values())
    savings = most_expensive_cost - cheapest_cost
    print(f"Вы сэкономите {savings} рублей, выбрав {cheapest_store}.")
```
