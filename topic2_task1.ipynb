```python
# Ввод слова
word = input("Введите слово: ")

# Проверяем длину слова
length = len(word)

# Если длина слова нечётная
if length % 2 != 0:
    # Выводим среднюю букву
    middle = word[length // 2]
    print("Средняя буква:", middle)
else:
    # Если длина слова чётная, выводим две средние буквы
    middle1 = word[length // 2 - 1]
    middle2 = word[length // 2]
    print("Средние буквы:", middle1 + middle2)
```
