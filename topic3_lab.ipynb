```python
from functools import lru_cache

@lru_cache(None)
def min_breaks(n, m):
    """
    Функция для определения минимального количества разломов шоколадки размером n x m.
    """
    # Если шоколадка уже 1x1, разломов не требуется
    if n == 1 and m == 1:
        return 0
    # Если одна из сторон равна 1, количество разломов равно длине другой стороны минус один
    elif n == 1:
        return m - 1
    elif m == 1:
        return n - 1
    # В общем случае: количество разломов равно количеству квадратиков минус один
    else:
        return n * m - 1

print(min_breaks(2, 3))  # 5
print(min_breaks(3, 3))  # 8
print(min_breaks(1, 1))  # 0
```
