```python
import json
import xml.etree.ElementTree as ET

# Чтение данных из JSON
def read_json(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        data = json.load(file)
    return data

# Конвертация данных в XML
def convert_to_xml(data):
    root = ET.Element("Люди")
    
    for person in data:
        person_element = ET.SubElement(root, "Человек")
        for key, value in person.items():
            child = ET.SubElement(person_element, key)
            child.text = str(value)
    
    return ET.ElementTree(root)

# Сохранение данных в XML
def save_to_xml(tree, output_path):
    tree.write(output_path, encoding='utf-8', xml_declaration=True)

# Основная программа
def main():
    input_file = "data.json"  # Исходный файл
    output_file = "data.xml"  # Файл для сохранения результата

    # Чтение данных из JSON
    data = read_json(input_file)

    # Конвертация в XML
    xml_tree = convert_to_xml(data)

    # Сохранение в XML
    save_to_xml(xml_tree, output_file)
    print(f"Данные успешно сохранены в файл {output_file}")

if __name__ == "__main__":
    main()
```
