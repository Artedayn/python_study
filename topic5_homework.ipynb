```python
from datetime import datetime

formats = {
    "The Moscow Times": "%A, %B %d, %Y",  # Wednesday, October 2, 2002
    "The Guardian": "%A, %d.%m.%y",       # Friday, 11.10.13
    "Daily News": "%A, %d %B %Y"          # Thursday, 18 August 1977
}

print("Введите данные в формате одной из газет или 'exit' для завершения:")

while True:
    try:
        user_input = input("Введите строку: ")
        
        if user_input.lower() == 'exit':
            print("Программа завершена.")
            break

        for name, date_format in formats.items():
            try:
                parsed_date = datetime.strptime(user_input, date_format)
                print(f"Газета: {name}, Дата: {parsed_date}")
                break
            except ValueError:
                continue
        else:
            print("Не удалось определить формат даты. Попробуйте снова.")
    
    except Exception as e:
        print(f"Произошла ошибка: {e}. Попробуйте снова.")
```
