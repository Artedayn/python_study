```python
import json

# Функция для загрузки данных о покупках из purchase_log.json
def load_purchase_data(purchase_file):
    with open(purchase_file, 'r', encoding='utf-8') as f:
        data = json.load(f)  # Загружаем весь JSON как список
    purchase_dict = {entry['user_id']: entry['category'] for entry in data}
    return purchase_dict

# Основная программа
def process_logs(visit_file, purchase_file, output_file):
    # Загружаем данные о покупках
    purchase_data = load_purchase_data(purchase_file)
    
    # Построчно читаем visit_log.csv и записываем в funnel.csv только те строки, где была покупка
    with open(visit_file, 'r', encoding='utf-8') as visit_f, \
         open(output_file, 'w', encoding='utf-8') as funnel_f:
        
        # Читаем заголовок из visit_log.csv
        header = visit_f.readline().strip()
        funnel_f.write(f"{header},category\n")  # Записываем заголовок в файл funnel.csv
        
        # Читаем строки файла visit_log.csv
        for line in visit_f:
            user_id, source = line.strip().split(',')
            if user_id in purchase_data:  # Если пользователь есть в данных о покупках
                category = purchase_data[user_id]
                funnel_f.write(f"{user_id},{source},{category}\n")  # Записываем строку в funnel.csv

# Указываем пути к файлам
visit_log_file = 'visit_log.csv'
purchase_log_file = 'purchase_log.json'  # Здесь JSON массив
output_funnel_file = 'funnel.csv'

# Запускаем обработку
process_logs(visit_log_file, purchase_log_file, output_funnel_file)
print(f"Файл {output_funnel_file} успешно создан!")
```
