```python
import time
from concurrent.futures import ThreadPoolExecutor

# Формула 1
def formula_1(x):
    return x**2 - x**2 + x**4 - x**5 + x + x

# Формула 2
def formula_2(x):
    return x + x

# Формула 3
def formula_3(result1, result2):
    return result1 + result2

# Функция для выполнения вычислений
def perform_calculations(iterations):
    start_time = time.time()
    
    # Результаты для формул 1 и 2
    results_1 = []
    results_2 = []
    
    # Параллельное выполнение формул 1 и 2
    with ThreadPoolExecutor() as executor:
        # Формула 1
        start_formula_1 = time.time()
        results_1 = list(executor.map(formula_1, range(iterations)))
        end_formula_1 = time.time()
        
        # Формула 2
        start_formula_2 = time.time()
        results_2 = list(executor.map(formula_2, range(iterations)))
        end_formula_2 = time.time()
    
    # Последовательное выполнение формулы 3
    start_formula_3 = time.time()
    results_3 = [formula_3(r1, r2) for r1, r2 in zip(results_1, results_2)]
    end_formula_3 = time.time()
    
    # Итоги времени выполнения
    total_time = end_formula_3 - start_time
    time_formula_1 = end_formula_1 - start_formula_1
    time_formula_2 = end_formula_2 - start_formula_2
    time_formula_3 = end_formula_3 - start_formula_3
    
    return time_formula_1, time_formula_2, time_formula_3, total_time

# Основная программа
if __name__ == "__main__":
    iterations_list = [10_000, 100_000]
    
    for iterations in iterations_list:
        print(f"Выполнение для {iterations} итераций:")
        time_formula_1, time_formula_2, time_formula_3, total_time = perform_calculations(iterations)
        
        print(f"  Время выполнения формулы 1: {time_formula_1:.6f} секунд")
        print(f"  Время выполнения формулы 2: {time_formula_2:.6f} секунд")
        print(f"  Время выполнения формулы 3: {time_formula_3:.6f} секунд")
        print(f"  Общее время выполнения: {total_time:.6f} секунд")
        print()
```
