```python
import re

def validate_car_number(car_id):
    # Регулярное выражение для проверки транспортного номера
    # Формат: 1 буква, 3 цифры, 2 буквы, 2–3 цифры
    pattern = r'^([АВЕКМНОРСТУХ])(\d{3})([АВЕКМНОРСТУХ]{2})(\d{2,3})$'
    
    match = re.match(pattern, car_id)
    if match:
        # Если номер валиден, извлекаем номер и регион
        number = match.group(1) + match.group(2) + match.group(3)
        region = match.group(4)
        return f"Номер {number} валиден. Регион: {region}."
    else:
        return "Номер не валиден."

# Примеры работы программы
car_id_1 = 'А222ВС96'
car_id_2 = 'АБ22ВВ193'

print(validate_car_number(car_id_1))  # Результат: Номер А222ВС валиден. Регион: 96.
print(validate_car_number(car_id_2))  # Результат: Номер не валиден.
```
